# AgentRobustBench: reproduce a paper result / 复现论文中的一个结果

**English.** Select **Runtime → Run all**. Keep the default CPU runtime. This notebook downloads the published score table and the calculation script, then recomputes how much ToolRL-DR-Full reduces the 3B ToolRL baseline's clean-to-Transition accuracy gap. No package installation, model download, API key, or GPU is needed.

**中文。** 选择菜单 **代码执行程序 → 全部运行（Runtime → Run all）**，使用默认 CPU 即可。本示例下载已公开的分数表和计算脚本，重新计算 ToolRL-DR-Full 相对 3B ToolRL 基线缩小了多少运行时错误造成的准确率差距。无需安装依赖、下载模型、填写 API key 或使用 GPU。

Expected output / 预期输出：

```text
ToolRL-Clean Transition gap:    31.49 pp
ToolRL-DR-Full Transition gap:  23.11 pp
Gap reduction:                  8.38 pp
Baseline gap closed:            26.6%
```

`pp` means percentage points / `pp` 表示百分点。

The calculation uses a fixed repository snapshot so that the input scores stay the same. To generate new model predictions, follow the [evaluation guide](https://github.com/WillChow66/AgentRobustBench/blob/main/scripts/README.md).

计算使用固定版本的仓库分数表。若要实际运行模型并生成新的预测，请查看[中文入门指南](https://github.com/WillChow66/AgentRobustBench/blob/main/README_zh-CN.md)。


````python
from pathlib import Path
from urllib.request import urlopen
import subprocess
import sys

# Fixed snapshot of the released calculation and scores.
revision = "72639db399e3972b215bab7116f2e03b8ae8b3f4"
base_url = f"https://raw.githubusercontent.com/WillChow66/AgentRobustBench/{revision}"
work_dir = Path("agentrobustbench-demo")

for relative_path in ("scripts/reproduce_transition_gap.py", "leaderboard/leaderboard.csv"):
    destination = work_dir / relative_path
    destination.parent.mkdir(parents=True, exist_ok=True)
    with urlopen(f"{base_url}/{relative_path}", timeout=30) as response:
        destination.write_bytes(response.read())

subprocess.run([sys.executable, str(work_dir / "scripts/reproduce_transition_gap.py")], check=True)
````
